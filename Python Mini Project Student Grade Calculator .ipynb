Problem Statement: Develop a Python program to calculate a student's grades, SGPA, and optional CGPA based on subject-wise marks and credits. 
The program should validate inputs, assign grades and grade points, determine the pass/fail status, and generate a formatted report card.

```python
# Student Grade Calculator

def calculate_grade(percentage):
    if percentage >= 90:
        return "A+"
    elif percentage >= 80:
        return "A"
    elif percentage >= 70:
        return "B"
    elif percentage >= 60:
        return "C"
    elif percentage >= 50:
        return "D"
    elif percentage >= 40:
        return "E"
    return "F"

def grade_point(g):
    return {"A+":10,"A":9,"B":8,"C":7,"D":6,"E":5,"F":0}[g]

print("="*80)
print("STUDENT GRADE CALCULATOR".center(80))
print("="*80)

student=input("Student Name : ")
roll=input("Roll Number  : ")
branch=input("Branch       : ")
cls=input("Class        : ")
semester=input("Semester     : ")

n=int(input("\nNumber of Subjects : "))

codes=[]; subjects=[]; obtained=[]; maximum=[]; grades=[]; gps=[]; credits=[]; cps=[]

for i in range(n):
    print(f"\nSubject {i+1}")
    code=input("Subject Code : ")
    sub=input("Subject Name : ")
    credit=int(input("Credits : "))

    tmax=float(input("Theory Maximum Marks : "))
    while True:
        t=float(input("Theory Obtained Marks : "))
        if 0<=t<=tmax: break
        print("Invalid marks! Enter again.")

    total=t
    max_total=tmax

    p=input("Practical Conducted? (Y/N): ").upper()
    if p=="Y":
        pmax=float(input("Practical Maximum Marks : "))
        while True:
            pm=float(input("Practical Obtained Marks : "))
            if 0<=pm<=pmax: break
            print("Invalid marks! Enter again.")
        total+=pm
        max_total+=pmax

    v=input("Viva Conducted? (Y/N): ").upper()
    if v=="Y":
        vmax=float(input("Viva Maximum Marks : "))
        while True:
            vm=float(input("Viva Obtained Marks : "))
            if 0<=vm<=vmax: break
            print("Invalid marks! Enter again.")
        total+=vm
        max_total+=vmax

    per=(total/max_total)*100
    g=calculate_grade(per)
    gp=grade_point(g)
    cp=gp*credit

    codes.append(code)
    subjects.append(sub)
    obtained.append(total)
    maximum.append(max_total)
    grades.append(g)
    gps.append(gp)
    credits.append(credit)
    cps.append(cp)

overall_obtained=sum(obtained)
overall_maximum=sum(maximum)
overall_percentage=(overall_obtained/overall_maximum)*100
overall_grade=calculate_grade(overall_percentage)
sgpa=sum(cps)/sum(credits)
result="PASS" if overall_percentage>=40 else "FAIL"

print("\n"+"="*120)
print("STUDENT REPORT CARD".center(120))
print("="*120)
print(f"Name      : {student}")
print(f"Roll No   : {roll}")
print(f"Branch    : {branch}")
print(f"Class     : {cls}")
print(f"Semester  : {semester}")
print("-"*120)
print(f"{'Code':<10}{'Subject':<22}{'Obt':<8}{'Max':<8}{'Grade':<8}{'GP':<6}{'Credits':<10}{'CP':<8}")
print("-"*120)
for i in range(n):
    print(f"{codes[i]:<10}{subjects[i]:<22}{obtained[i]:<8.1f}{maximum[i]:<8.1f}{grades[i]:<8}{gps[i]:<6}{credits[i]:<10}{cps[i]:<8}")
print("-"*120)
print(f"Total Marks : {overall_obtained:.1f}/{overall_maximum:.1f}")
print(f"Percentage  : {overall_percentage:.2f}%")
print(f"Overall Grade : {overall_grade}")
print(f"SGPA : {sgpa:.2f}")
print(f"Result : {result}")

ch=input("\nCalculate CGPA? (Y/N): ").upper()
if ch=="Y":
    prev=int(input("Number of Previous Semesters : "))
    total=sgpa
    for i in range(prev):
        total+=float(input(f"Semester {i+1} SGPA : "))
    print(f"CGPA : {total/(prev+1):.2f}")

print("="*120)
print("THANK YOU".center(120))
print("="*120)
```
